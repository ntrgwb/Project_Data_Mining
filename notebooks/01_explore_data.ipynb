```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score

train_df = pd.read_csv('dataset/dataset/train.csv')
dev_df = pd.read_csv('dataset/dataset/dev.csv')

# 2. HÀM TRÍCH XUẤT ĐẶC TRƯNG ABSA
aspect_keywords = {
    'aspect_quality': ['chất lượng', 'xịn', 'tốt', 'vải', 'đẹp', 'bền', 'dỏm', 'kém'],
    'aspect_delivery': ['giao hàng', 'vận chuyển', 'ship', 'nhanh', 'chậm', 'đóng gói'],
    'aspect_price': ['giá', 'rẻ', 'đắt', 'tiền', 'voucher', 'xu']
}

def extract_absa_features(text):
    text_lower = str(text).lower()
    features = []
    for aspect, keywords in aspect_keywords.items():
        has_aspect = int(any(kw in text_lower for kw in keywords))
        features.append(has_aspect)
    return features

# 3. TIỀN XỬ LÝ VÀ CHỐNG RÒ RỈ DỮ LIỆU (DATA LEAKAGE)
# Chỉ fit_transform trên tập TRAIN
tfidf = TfidfVectorizer(max_features=500)
X_train_text = tfidf.fit_transform(train_df['Comment']).toarray()
print("X_train_text")
print(X_train_text)
print("==============================")
X_train_rating = (train_df[['Rating']].values - 1) / 4.0
absa_train = np.array([extract_absa_features(c) for c in train_df['Comment']])

X_train = np.hstack([X_train_text, X_train_rating, absa_train])
print("X_train")
print(X_train)
print("==============================")
y_train = train_df['Label'].values
print("y_train", y_train)
print("==============================")
print("absa_train")
print(absa_train)
print("==============================")



# Chỉ transform trên tập DEV (Validation)
X_dev_text = tfidf.transform(dev_df['Comment']).toarray()
X_dev_rating = (dev_df[['Rating']].values - 1) / 4.0
absa_dev = np.array([extract_absa_features(c) for c in dev_df['Comment']])

X_val = np.hstack([X_dev_text, X_dev_rating, absa_dev])
y_val = dev_df['Label'].values

print("Kích thước ma trận huấn luyện (X_train):", X_train.shape)
print("Kích thước ma trận kiểm tra (X_val):", X_val.shape)

# 4. HUẤN LUYỆN MÔ HÌNH CƠ SỞ (BASELINE)
clf = LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42)
clf.fit(X_train, y_train)

# 5. ĐÁNH GIÁ TRÊN TẬP DEV
y_pred = clf.predict(X_val)
print("du doan", y_pred)
print("That   ", y_val)


print("\n--- BÁO CÁO PHÂN LOẠI TRÊN TẬP DEV (CLASSIFICATION REPORT) ---")
print(classification_report(y_val, y_pred, target_names=['0: Non-Spam', '1: Spam']))
print("Macro F1-Score:", round(f1_score(y_val, y_pred, average='macro'), 4))
```
