```python
import pandas as pd
train_df = pd.read_csv("../dataset/train.csv")
val_df = pd.read_csv("../dataset/dev.csv")
test_df = pd.read_csv("../dataset/test.csv")

print("Kich thuoc tep train, dev, test:", train_df.shape, val_df.shape, test_df.shape)
print("Danh sách các cột trong tệp train, dev, test:", list(train_df.columns))
print("="*10)
print("Các dòng đầu tiên của dữ liệu")
print(train_df.head(3))
print("="*10)
print("Kiểm tra null của tệp train")    
print(train_df.isnull().sum())
print("Kiểm tra giá trị trùng lậ    p")
print(train_df.duplicated().sum())

#Phân bố Rating theo nhãn spam/non-spam
print("Phân bố rating theo nhãn")
rating_label = pd.crosstab(train_df['Rating'], train_df['Label'], margins=True)
print(rating_label)

```
